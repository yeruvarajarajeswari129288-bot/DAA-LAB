```python
import time

def coin_change(coins, amount):
    # Create DP array
    dp = [float('inf')] * (amount + 1)

    # Base case
    dp[0] = 0

    # Fill DP array
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)

    if dp[amount] == float('inf'):
        return -1

    return dp[amount]


# Input
coins = list(map(int, input("Enter coin denominations: ").split()))
amount = int(input("Enter the amount: "))

# Start execution time
start_time = time.perf_counter()

# Calculate minimum coins
result = coin_change(coins, amount)

# End execution time
end_time = time.perf_counter()

# Calculate execution time
execution_time = end_time - start_time

# Output
if result == -1:
    print("Change cannot be made")
else:
    print("Minimum number of coins =", result)

print("Execution Time =", execution_time, "seconds")
print("Time Complexity = O(n × amount)")
print("Space Complexity = O(amount)")
```
