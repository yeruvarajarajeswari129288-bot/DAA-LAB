```python
# 1.LINEAR SEARCH
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1


# Input
while True:
    try:
        arr_input = input("Enter elements (space-separated): ")
        arr = list(map(int, arr_input.split()))
        break
    except ValueError:
        print("Invalid input. Please enter space-separated integers.")


# Key input
while True:
    try:
        key = int(input("Enter element to search: "))
        break
    except ValueError:
        print("Invalid input. Please enter a single integer.")


# Search
result = linear_search(arr, key)


# Output
if result != -1:
    print("\nElement found at index:", result)
else:
    print("\nElement not found")


# Time Complexity
print("\n--- Time Complexity ---")
print("Best Case    : O(1)")
print("Average Case : O(n)")
print("Worst Case   : O(n)")

# Space Complexity
print("\nSpace Complexity: O(n)")
```

```python
# 2.BINARY_search
def binary_search(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            return mid

        elif arr[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    return -1


# Input
while True:
    try:
        arr_input = input("Enter sorted elements (space-separated): ")
        arr = list(map(int, arr_input.split()))

        # Check whether array is sorted
        if arr != sorted(arr):
            print("Please enter elements in sorted order.")
            continue

        break

    except ValueError:
        print("Invalid input. Please enter space-separated integers.")


# Key input
while True:
    try:
        key = int(input("Enter element to search: "))
        break
    except ValueError:
        print("Invalid input. Please enter a single integer.")


# Search
result = binary_search(arr, key)


# Output
if result != -1:
    print("\nElement found at index:", result)
else:
    print("\nElement not found")


# Time Complexity
print("\n--- Time Complexity ---")
print("Best Case    : O(1)")
print("Average Case : O(log n)")
print("Worst Case   : O(log n)")

# Space Complexity
print("\nSpace Complexity: O(1)")
```
